```python
cols_withna = [col for col in df.columns if df[col].isnull().sum() > 0]
```

```python
df.shape
```

```python
from sklearn.model_selection import train_test_split
X_train , X_test , y_train , y_test = train_test_split(
    df.drop('MEDV' , axis=1),
    df['MEDV'],
    test_size=0.2,
    random_state=42
)
```

```python
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(strategy='mean')
X_train[cols_withna] = imputer.fit_transform(X_train[cols_withna])
X_test[cols_withna] = imputer.transform(X_test[cols_withna])
```

```python
# train a random forest model
from sklearn.ensemble import RandomForestRegressor
rf = RandomForestRegressor(n_estimators= 200 , max_depth=300, verbose=True , n_jobs=-1)
rf.fit(X_train , y_train)
rf.score(X_test , y_test)
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(X_train , y_train)
lr.score(X_test , y_test)
```

```python
from sklearn.linear_model import SGDRegressor
sgd = SGDRegressor()
sgd.fit(X_train , y_train)
sgd.score(X_test , y_test)
```

```python
from sklearn.ensemble import GradientBoostingRegressor
gb = GradientBoostingRegressor(n_estimators=300)
gb.fit(X_train , y_train)
gb.score(X_test , y_test)
```

```python
# A good starting parameter grid for RandomForestClassifier
rf_param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [None, 10, 20],
    'min_samples_split': [ None , 2, 5, 10],
    'min_samples_leaf': [None , 1, 2, 4],
    'max_features': [None , 'sqrt'] # A common and effective choice
}
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import GridSearchCV

# 1. Instantiate the model
# Using random_state for reproducibility
rf = RandomForestRegressor(random_state=42)

# 2. Set up GridSearchCV
# cv=5 means 5-fold cross-validation
# n_jobs=-1 uses all available CPU cores
# verbose=2 prints progress updates
grid_search_rf = GridSearchCV(estimator=rf,
                              param_grid=rf_param_grid,
                              cv=5,
                              n_jobs=-1,
                              verbose=2)

# 3. Fit the grid search to your data (assuming X_train, y_train exist)
grid_search_rf.fit(X_train, y_train)

# After fitting, you can find the best parameters and score
print(f"Best Parameters for Random Forest: {grid_search_rf.best_params_}")
print(f"Best Cross-validation Score: {grid_search_rf.best_score_:.4f}")
```

```python
# A good starting parameter grid for GradientBoostingClassifier
gb_param_grid = {
    'n_estimators': [400 , 500 , 600 , 700],
    'learning_rate': [0.01, 0.1, 0.2],
    'max_depth': [3, 5, 7],
    'subsample': [0.8, 1.0] # Fraction of samples used to train each tree
}

from sklearn.ensemble import GradientBoostingRegressor
from sklearn.model_selection import GridSearchCV

# 1. Instantiate the model
gb = GradientBoostingRegressor(random_state=42)

# 2. Set up GridSearchCV
grid_search_gb = GridSearchCV(estimator=gb,
                              param_grid=gb_param_grid,
                              cv=5,
                              n_jobs=-1,
                              verbose=2)

# 3. Fit the grid search to your data
grid_search_gb.fit(X_train, y_train)

# After fitting, find the best parameters and score
print(f"Best Parameters for Gradient Boosting: {grid_search_gb.best_params_}")
print(f"Best Cross-validation Score: {grid_search_gb.best_score_:.4f}")
```
